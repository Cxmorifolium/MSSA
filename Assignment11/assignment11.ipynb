## 121. Best Time to Buy and Sell Stock

```You are given an array prices where prices[i] is the price of a given stock on the ith day.

You want to maximize your profit by choosing a single day to buy one stock and choosing a different day in the future to sell that stock.

Return the maximum profit you can achieve from this transaction. If you cannot achieve any profit, return 0.

Example 1:

Input: prices = [7,1,5,3,6,4]
Output: 5
Explanation: Buy on day 2 (price = 1) and sell on day 5 (price = 6), profit = 6-1 = 5.
Note that buying on day 2 and selling on day 1 is not allowed because you must buy before you sell.

Example 2:

Input: prices = [7,6,4,3,1]
Output: 0
Explanation: In this case, no transactions are done and the max profit = 0.```



````python
// Similar to a previous problem
public class Solution
{
    public int MaxProfit(int[] prices)
    {
        int min = int.MaxValue; // it is at some value
        int max = 0;

        for (int i = 0; i < prices.Length; i++)
        {
            min = Math.Min(min, prices[i]); // Set current min to the initial index of prices[] comparing price[i] and min (when updated)

            int profit = prices[i] - min; // profit = current price - current min; 

            max = Math.Max(max, profit); // update max and compare between current profit and max profit
        }

        return max;
    }
}
````

## 206. Reversed Linked List

Given the head of a singly linked list, reverse the list, and return the reversed list.

````python
public class Solution {public ListNode ReverseList(ListNode head) 
    {
        ListNode prev = null; // becuase null is at the end of the list
        ListNode current = head;
        
        while (current != null) {
            // Save where pointer is going to go next
            ListNode next = current.next;
            
            // Point current backwards
            current.next = prev;
            
            // Move both pointers forward (er backwards?)
            prev = current;
            current = next;
        }
        
        return prev;  // prev is now the new head
    }
}
````
