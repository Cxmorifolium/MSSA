## 383. Ransom Note

Given two strings ransomNote and magazine, return true if ransomNote can be constructed by using the letters from magazine and false otherwise.

Each letter in magazine can only be used once in ransomNote.

 

Example 1:

Input: ransomNote = "a", magazine = "b"
Output: false

Example 2:

Input: ransomNote = "aa", magazine = "ab"
Output: false

Example 3:

Input: ransomNote = "aa", magazine = "aab"
Output: true


```python
public class Solution {
    public bool CanConstruct(string ransomNote, string magazine) 
    {
        Dictionary<char, int> letterCounts = new Dictionary<char, int>();

        foreach (char letter in magazine)
        {
            if (letterCounts.ContainsKey(letter))
                letterCounts[letter]++; 
            else
                letterCounts[letter] = 1;
        }

        foreach (char letter in ransomNote)
        {
            // Check if letter doesn't exist OR count is 0
            if (!letterCounts.ContainsKey(letter) || letterCounts[letter] == 0)
                return false;
            
            letterCounts[letter]--;  // decrease the count
        }
        
        return true;
    }
}

// Time Complexity O(n)
```

## 234. Palindrome Linked List

Given the head of a singly linked list, return true if it is a

or false otherwise.

 

Example 1:

Input: head = [1,2,2,1]
Output: true

Example 2:

Input: head = [1,2]
Output: false


```python
/**
 * Definition for singly-linked list.
 * public class ListNode {
 *     public int val;
 *     public ListNode next;
 *     public ListNode(int val=0, ListNode next=null) {
 *         this.val = val;
 *         this.next = next;
 *     }
 * }

 */
public class Solution {
    public bool IsPalindrome(ListNode head)
    {
        Stack<int> stack = new Stack<int>();
        ListNode current = head;  // moves one step at a time
        ListNode runner = head;   // moves 2 steps at a time to find middle
        
        while (runner != null && runner.next != null)
        {
            stack.Push(current.val);
            current = current.next;
            runner = runner.next.next;
        }
        
        if (runner != null)  // odd length
        {
            current = current.next;
        }
        
        while (current != null)
        {
            if (stack.Pop() != current.val)
            { 
                return false;
            }
            current = current.next;
        }
        
        return true;
    }

}

// O(n) complexity
```

# 203 Removed Linked List Elements

Given the head of a linked list and an integer val, remove all the nodes of the linked list that has Node.val == val, and return the new head. <break>

Example 1:<break>

Input: head = [1,2,6,3,4,5,6], val = 6  
Output: [1,2,3,4,5]  


```python
/**
 * Definition for singly-linked list.
 * public class ListNode {
 *     public int val;
 *     public ListNode next;
 *     public ListNode(int val=0, ListNode next=null) {
 *         this.val = val;
 *         this.next = next;
 *     }
 * }
 */
public class Solution {
    public ListNode RemoveElements(ListNode head, int val) {
    ListNode dummy = new ListNode(0);
    dummy.next = head;

    ListNode prev = dummy;
    ListNode current = head;

    while (current != null)
    {
        if (current.val == val)
        {
            prev.next = current.next;
        }
        else
        {
            prev = current;
        }

        current = current.next;
    }

    return dummy.next;
    }

}

// Time: O(n)
```

# 1047. Remove All Adjacent Duplicates In String

You are given a string s consisting of lowercase English letters. A duplicate removal consists of choosing two adjacent and equal letters and removing them. <break>

We repeatedly make duplicate removals on s until we no longer can. <break>

Return the final string after all such duplicate removals have been made. It can be proven that the answer is unique.
 

Example 1: <break>

Input: s = "abbaca"  
Output: "ca"  
Explanation:   
For example, in "abbaca" we could remove "bb" since the letters are adjacent and equal, and this is the only possible move.  The result of this move is that the string is "aaca", of which only "aa" is possible, so the final string is "ca".


```python
public class Solution {
    public string RemoveDuplicates(string s) {
        StringBuilder sb = new StringBuilder();

        foreach (char c in s)
        {
            if (sb.Length > 0 && sb[sb.Length-1] ==c)
            {
                sb.Length--;
            }
            else
            {
                sb.Append(c);
            }
        }

        return sb.ToString();   
    }
}
```
