1. Write a program in C# Sharp to find the +ve numbers from a list of numbers using where conditions in LINQ Query.
    ```
    Input: { 2, -1, 3, -3, 10, -200}
    Expected output: {2, 3, 10}
    ```

````python
using System.Collections.Generic;
using System.Linq;
using System;
````

````python
List<int> numbers = new List <int> {2, -1, 3, -3, 10, -200};
        
//Linq
var positiveNumbers = from num in numbers
                        where num > 0
                        orderby num ascending
                        select num;

Console.WriteLine($"List of numbers: {string.Join(", ", numbers)}");
Console.WriteLine($"Positive Numbers: {string.Join(", ", positiveNumbers)}");
````

2. Write a program to create a list of employees. Consider a hard coded list. Display all employees who have salary more than $5000 and age < 30.

````python
// Employee class with properties
public class Employee
{
    public string Name { get; set; }
    public int Age { get; set; }
    public decimal Salary { get; set; }
    
    // Constructor
    public Employee(string name, int age, decimal salary)
    {
        Name = name;
        Age = age;
        Salary = salary;
    }
    
    // ToString for display
    public override string ToString()
    {
        return $"{Name}, Age: {Age}, Salary: ${Salary:N0}";
    }
}

List<Employee> employees = new List<Employee>
{
    new Employee("Alex", 25, 6000),
    new Employee("Barbie", 35, 7500),
    new Employee("Catherine", 48, 4500),
    new Employee("Dahlia", 55, 5500),
    new Employee("Elphie", 65, 8000),
    new Employee("Fiona", 55, 9500),
    new Employee("Georgia", 45, 10500)
};

Console.WriteLine("List of Employees: ");

foreach(var emp in employees)
{
    Console.WriteLine(emp);
}

// LINQ query to filter employees with salary > $5000 and age < 30
var filteredEmployees = from emp in employees
                       where emp.Salary > 5000 && emp.Age < 30
                       select emp;

Console.WriteLine("\nEmployees with salary > $5000 and age < 30:");
foreach(var emp in filteredEmployees)
{
    Console.WriteLine(emp);
}
````

3. Write a program in C# Sharp to find a string that starts and ends with a specific character. <break>
    ```
    Test Data : 
    The cities are : 'ROME','LONDON','NAIROBI','CALIFORNIA','ZURICH','NEW DELHI','AMSTERDAM','ABU DHABI','PARIS' 
    Input starting character for the string : A
    Input ending character for the string : M 
    ```

````python
List<string> cities = new List<string> {"ROME", "LONDON", "NAIROBI", "CALIFORNIA", "ZURICH", "NEW DELHI", "AMSTERDAM", "ABU DHABI", "PARIS"};
var filteredCities = from city in cities
                    where city.StartsWith("A") && city.EndsWith("M")
                    select city;

Console.WriteLine($"List of Cities: {string.Join(", ", cities)}");
Console.WriteLine($"Cities starting with A and ending with M: {string.Join(", ", filteredCities)}");
````

Write a program in C# Sharp to create a list of numbers and display numbers greater than 80. <break>

```
    Test Data :
    The members of the list are :
    55 200 740 76 230 482 95
    Expected Output :
    The numbers greater than 80 are :
    200
    740
    230
    482
    95
```

````python
List<int> numberlist = new List<int> {55, 200, 740, 76, 230, 482, 95};

var numberGreaterThan = from nums in numberlist
                        where nums > 80
                        select nums;
                        
Console.WriteLine($"Numbers list: {string.Join(", ", numberlist)}");
Console.WriteLine($"Numbers greater than 80: {string.Join(", ", numberGreaterThan)}");
````
