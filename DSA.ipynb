#### Recursion

The code below uses nested loops to print each directory and its subdirectories up to multiple levels deep.

````python
using System.IO;

class Program
{
    static void Main(string[] args)
    {
        const string path = @"C:\"; // path here
        string[] directories = Directory.GetDirectories(path);
        foreach (var d in directories)
        {
            Console.WriteLine(d);
            string[] subdirs = Directory.GetDirectories(d);
            foreach(var subdir in subdirs)
            {
                Console.WriteLine(subdir);
                string[] subsubdirs = Directory.GetDirectories(subdir);
                foreach(var subsubdir in subsubdirs)
                {
                    Console.WriteLine(subsubdir);
                }
            }
        } 
        Console.ReadKey();
    }
}
````

So where does recursion come in? Recursion allows the program to explore directories to an arbitrary depth without needing to manually add more nested loops — it simply calls the same method for each subdirectory, continuing the process until there are no more.

+ Recursion: a technique to solve reptition logic by calling the function itself.

Suppose we have:

````python
void A()
{
    A();
}
````

Here, function A is calling itself with no condition to stop — this results in infinite recursion. If there's no base case (a stopping point), the program will eventually crash with a stack overflow error, because each function call consumes memory on the call stack.


So what happens in memory?

Let’s say we have a Main method that calls two functions: Add() and Product().

When the program runs:

+ The Main method is pushed onto the call stack.

+ Any local variables declared in Main are also stored in this stack frame.

+ When Main calls Add(), the Add function is pushed onto the stack, above Main.

    + The stack now holds both Main and Add.

    + Local variables in Add, and any operations inside it, are handled in its own stack frame.

+ Once Add() finishes executing and returns a value, its frame is popped off the stack, and the memory is reclaimed.

+ The program continues in Main, and now calls Product().

    + Product is pushed onto the stack, and works the same way.

    + When Product() finishes, it's popped off the stack as well.

+ This push-pop behavior of the call stack ensures that memory is properly managed — but if a function (like A) keeps calling itself with no exit, the stack keeps growing until it overflows-- throwing an exception.

So lets rewrite it:

````python
using System.IO;

class Program 
{
    // Recursion method
    static void PrintDirectories(string path, int depth)
    {
        string[]directories = Directory.GetDirectories(path);
        foreach (var d in directories)
        {
            Console.WriteLine(d);
            if(depth>0)
            {
                PrintDirectories(d, depth-1); // Exit condition
            }
        }
    }

    // Main program: call it here
    static void Main(string[] args)
    {
        const string path =@"C:\"; // path here
        string[] directories=Directory.GetDirectories(path);
        PrintDirectories(path, 2);
    }
}
````

1. Recursive Method (PrintDirectories):

    + Takes a path and a depth limit.

    + Lists all subdirectories in the current path and prints them.

    + If the depth is greater than 0, it calls itself on each subdirectory, decreasing the depth by 1 (this is the exit condition).

2. Base Case (Exit Condition):

    + The recursion stops when depth reaches 0, preventing infinite recursion and limiting how deep the directory structure is explored.

3. Main Method:

    + Sets the starting path to C:\.

    + Calls PrintDirectories with a depth of 2, meaning it will print directories up to 2 levels deep from the root.

````python
using System;

class Program
{
    // Tail Recursion: Recursive call is the last thing done
    static void Square_TailRecursion(int n)
    {
        if(n>0)
        {
            Console.WriteLine(n*n);
            Square_TailRecursion(n-1);
        }
    }

    // Head Recursive: Recursive call is the first thing done
    // Print in reverse order
    static void Square_HeadRecursion(int n)
    {
        if (n>0)
        {
            Square_HeadRecursion(n-1);
            Console.WriteLine(n*n);
        }
    }

    static void Main(string[] args)
    {
        Square_HeadRecursion(3);
        Square_TailRecursion(3);
    }
}

//Output head: 1 4 9
//Output tail : 9 4 1
````

##### Recursion Example 3:

Problem statement: Calculate the sum of all the numbers 1 to n.

````python
using System;

class Program{
    
    // By Iteration
    static int SumIteration(int n)
    {
        int sum = 0;
        int i = 1;
        while (i<=n)
        {
            sum = sum+i;
            i++;
        }
        return sum;
    }

    // By Recursion
    // Mathmatically:
    // n = 5: 1+2+3+4+5 == sum(4)+5
    // n = 5: 1+2+3+4+5+6 == sum(5)+6
    // sum(n) = sum(n-1)+n

    static int SumRecursion(int n)
    {
        if(n==0) return 0; // base condition
        if(n==1) return 1;
        return SumRecursion(n-1) + n;

    }

    static void Main(string[] args)
    {
        int num = 4;
        Console.WriteLine($"Num: {num}; {SumRecursion(num)}");
    }
}
````

##### 1. Brute Force Algorithm
+ Definition: Tries all possible solutions until it finds the correct one.
+ Pros: Simple to implement; guarantees a correct solution.
+ Cons: Very slow for large inputs.

````python
static bool BruteForceSearch(int[] arr, int target)
    {
        foreach (int num in arr)
        {
            if (num == target)
                return true;
        }
        return false;
    }
````

````python
static int Factorial(int n)
    {
        if (n == 0)
            return 1;
        return n * Factorial(n - 1);
    }
````

##### 3. Backtracking Algorithm
+ Definition: Tries solutions and abandons them if they don't work.
+ Pros: Useful for constraint problems (e.g. Sudoku, N-Queens).
+ Cons: Can be slow; needs careful pruning.

````python
// Backtracking Algorithm (Simple N-Queens check for size 4)
    static bool SolveNQueens(int n)
    {
        List<(int row, int col)> board = new List<(int, int)>();

        bool IsSafe(int row, int col)
        {
            foreach (var (r, c) in board)
            {
                if (c == col || Math.Abs(r - row) == Math.Abs(c - col))
                    return false;
            }
            return true;
        }

        bool Backtrack(int row)
        {
            if (row == n)
                return true;
            for (int col = 0; col < n; col++)
            {
                if (IsSafe(row, col))
                {
                    board.Add((row, col));
                    if (Backtrack(row + 1))
                        return true;
                    board.RemoveAt(board.Count - 1);
                }
            }
            return false;
        }

        return Backtrack(0);
    }
````

##### 4. Searching and Sorting Algorithms
+ Definition: Techniques to find data (searching) or arrange data (sorting).
+ Pros: Essential for optimizing data use; many options like binary search or quicksort.
+ Cons: Some are inefficient for large or unsorted data.



````python
static bool BinarySearch(int[] arr, int target)
    {
        int low = 0, high = arr.Length - 1;
        while (low <= high)
        {
            int mid = (low + high) / 2;
            if (arr[mid] == target)
                return true;
            else if (arr[mid] < target)
                low = mid + 1;
            else
                high = mid - 1;
        }
        return false;
    }

static void BubbleSort(int[] arr)
    {
        for (int i = 0; i < arr.Length; i++)
        {
            for (int j = 0; j < arr.Length - 1 - i; j++)
            {
                if (arr[j] > arr[j + 1])
                {
                    int temp = arr[j];
                    arr[j] = arr[j + 1];
                    arr[j + 1] = temp;
                }
            }
        }
    }
````

##### 5. Divide-and-Conquer (DAC) Algorithms
+ Definition: Breaks problems into smaller pieces, solves them, and combines the results.
+ Pros: Fast and efficient (e.g. mergesort, quicksort).
+ Cons: Can be hard to design; recursion overhead.



````python
static int[] MergeSort(int[] arr)
    {
        if (arr.Length <= 1)
            return arr;

        int mid = arr.Length / 2;
        int[] left = MergeSort(arr[..mid]);
        int[] right = MergeSort(arr[mid..]);

        return Merge(left, right);
    }

    static int[] Merge(int[] left, int[] right)
    {
        List<int> result = new List<int>();
        int i = 0, j = 0;

        while (i < left.Length && j < right.Length)
        {
            if (left[i] < right[j])
                result.Add(left[i++]);
            else
                result.Add(right[j++]);
        }

        while (i < left.Length) result.Add(left[i++]);
        while (j < right.Length) result.Add(right[j++]);

        return result.ToArray();
    }
````

##### 7. Greedy Algorithms
+ Definition: Makes the best local choice at each step.
+ Pros: Simple and fast; good for optimization (e.g. Kruskal’s).
+ Cons: Doesn't always give the best global solution.

````python
static int GreedyCoinChange(int[] coins, int amount)
    {
        Array.Sort(coins);
        Array.Reverse(coins);
        int count = 0;

        foreach (int coin in coins)
        {
            while (amount >= coin)
            {
                amount -= coin;
                count++;
            }
        }

        return amount == 0 ? count : -1; // Return -1 if no solution
    }
````

##### Binary Search



Divide and Conquer apporach and works with sorted list

Algorithm

The binary search uses the midpoint element and then proceeds with the search. It uses the following algorithm:

- Step 1: You identify the element you need to search for.
- Step 2: You then select a midpoint in the list. One of the elements in the list is the midpoint.
- Step 3: You check whether the searched elements are smaller or larger than the midpoint.
- Step 4: You move to a new midpoint based on whether it is smaller or larger than the previous midpoint. If the searched element is smaller, the search moves to the elements before the midpoint.
- Step 5: You need to search by only considering the elements before the midpoint.

Time Space Complexity:
- Best: O(1) searched element is at the midpoint, which is the starting point
- Avg: O(log n)
- Worst: O(log n)

### Queue


Queue is a lienar DS that collects similar data items

-Deletion: Dequeue()

````python
using System.Collections.Generic;

class Q
{
    private int[] data;
    private int front, rear, size;

    public Q(int size)
    {
        this.data = new int[size]; //max number in array
        this.front = -1;
        this.rear = -1;
        this.size = 0;
    }

    public bool IsEmpty()
    {
        return this.size == 0;
    }

    public bool IsFull()
    {
        return this.rear == this.data.Length-1;
    }

    public void Enqueue(int val)
    {
        if(IsFull())
        {
            Console.WriteLine("Q is full. Cannot add");
            return;
        }
        else
        {   //Out of bounds check
            if(rear < this.data.Length-1)
                rear++;
            data[rear] = val;
            size++;
        }
    }

    public int Dequeue()
    {
        if(IsEmpty())
        {
            Console.WriteLine("Q is empty!");
            return -1;
        }
        else
        {
            front++;
            size --;
            int temp = data[front];
            if(IsEmpty())
            {
                this.front = -1;
                this.rear = -1;
            }
            return temp;
        }
    }

    public void Display()
    {
        if(!IsEmpty())
        {
            for( int i = front+1; i <= rear; i ++)
            {
                Console.Write(data[i] + " ");
            }
        }
        else
            Console.WriteLine("Q is empty.");
            Console.WriteLine();
    }

    static void Main (string[] args)
    {
        Q myQ = new Q(8);
        myQ.Enqueue(1);
        myQ.Enqueue(2);
        myQ.Enqueue(3);
        myQ.Enqueue(4);
        myQ.Display();
        
        // test methods here
    }

}
````

````python
using System.Collections.Generic;

class Node
{
    public int data;
    public Node next;
    public Node(int val, Node next)
    {
        this.data = val;
        this.next = next;
    }

}

class Q
{
    Node front; //head
    Node rear; //tail
    int size; 
    public bool IsEmpty()
    {
        return this.size == 0;
    }
    public void Enqueue(int val)
    {
        Node newNode = new Node(val, null);
        if (IsEmpty())
        {
            front = rear = newNode;
        }
        else
        {
            rear.next = newNode;
            rear = newNode;
        }
        size++;
    }

    public int? Dequeue()
    {
        if(IsEmpty()) 
        {
            Console.WriteLine("Q is empty");
            return null;
            
        }
        int val = front.data;
        front = front.next;
        size --;
        if (IsEmpty())
            rear = null;
        return val;
    }

    public void Display()
    {
        Node temp = front;
        if(!IsEmpty())
        {
            while (temp != null)
            {
                Console.Write(temp.data + " ");
                temp = temp.next;
            }
        }
        else
            Console.WriteLine("Q is empty");
    }

    static void Main(string[] args)
    {
        Q myq = new Q();
        myq.Enqueue(12);
        myq.Enqueue(13);
        myq.Display();
        myq.Dequeue();
        Console.ReadKey();
    }
}

````

### Sorting Techniques

#### Bubble Sort

1. Compare consecutive elements
2. If left side element is greater than right, swap them
    - [ 20, 11 ] => [ 11, 20 ] throughout the array
3. Continue this process till the end of the array and perform several rounds/passes
<break>

Consider: [ 3 5 8 9 6 2]  
3 < 5 < 8 < 9 > 6 > 2 = Swap needed  
<break>

Round 1: 3 5 8 6 2 9; 9 is sorted  
Round 2: 3 5 6 8 2 9  
Round 3: 3 5 6 2 8 9; now 8 is sorted  
Round 4: 3 5 2 6 8 9; 6 is sorted  
Round 5: 3 2 5 6 8 9; 5 is sorted  
Round 6: 2 3 5 6 8 9; completed 
<break>

Stable Algorithm: A stable algorithm maintains the relative order of elements with equal keys (i.e., duplicates). This means that if two elements are equal, the one that appears first in the input will also appear first in the output
<break>

Consider: [ 23 2 14 1 2]
Round 1: 2 23 14 1 2  
Round 2: 2 14 23 1 2  
Round 3: 2 14 1 23 2  
Round 4: 2 14 1 2 23 -> 23 is sorted  
Round 5: 2 1 14 2 23  
Round 6: 2 1 2 14 23 -> 14 is sorted  
Round 7: 1 2 2 14 23 -> completed

````python
// TC: O(n^2)
static void BubbleSort(int[] arr)
{
    int temp = 0;
    for (int pass = arr.Length-1; pass >=0; pass --)
    {
        for (int i = 0; i < pass; i ++)
        {
            if (arr[i] > arr[i+1])
            {
                temp = arr[i];
                arr[i] = arr[i+1];
                arr[i+1] = temp;
            }
        }
    }
}

static void Main(string[] args)
{
    int[] A = {23, 12, 11, 2, 45, 100};
    BubbleSort(A);
    Console.WriteLine("Sorted Array: ");
    foreach(int i in A)
    {
        Console.WriteLine(i);
    }
    Console.ReadKey();
}

//Output: 2 11 12 23 45 100
````

#### Selection Sort:

Not very efficient:
1. Find the smallest element in an array and replace it with the first element.
2. Find the smallest element in the unsorted part of an array and replace it with the second element.
3. Repeat this process until the array is sorted at the end.

Advantages
1. It is easy to implement.
2. It works well on small lists.
3. It does not require additional temporary storage.
<break>

Disadvantages of Selection Sort
1. It does not work well for a large number of elements.
2. It is not stable.

Time Complexity best and worst O(N^2)  
Space complexity O(1)
<break>

Unstable algorithm

````python
// 23 2 11
// i = 0 = 23; j =  i+1 = 2
// is 23 < 2?
// inner loop find min number
// outer tracks position

// Manaul debug: [ 3 5 8 9 6 2]
// min position = 0; i = 0; j = 1;
// if ( 5 < 3 ) => no, do nothing
// if ( 8 < 3 ) = " "
// if ( 9 < 3 ) = " "
// if ( 6 < 3 ) = " "
// if ( 2 < 3 ) =  minPosition = [5]; swap place
// [ 2 5 8 9 6 3 ]
// i = 1; min position = 1; j = 2
// if ( 8 < 5 ); if ( 9 < 5)...if( 5 < 3) swap 
// [ 2 3 8 9 6 5] etc
// then it goes to 2nd to last element

static void SelectionSort(int[]A)
{
    int minPosition = 0; //l to r
    int temp = 0;

    // i is tracking the correct position to put the element
    for (int i = 0; i < A.Length-1; i++)
    {
        // minPosition tracks the index of the min number
        minPosition = i;
        for (int j = i+1; j <A.Length; j++)
        {
            if(A[j] < A[minPosition]) // logic to find min number
            {
                minPosition = j; // update the minPosition with index of min number
            }
        }

        if(minPosition != i)
        {
            temp = A[i];
            A[i] = A[minPosition];
            A[minPosition] = temp;
        }
    }
}

static void Main(string[] args)
{
    int[] arr = { 3, 5, 8, 9, 6, 2 };
    SelectionSort(arr);
    Console.WriteLine("sorted array: ");
    foreach(int i in arr) Console.Write(i + " ");
}
````

#### Insertion Sort

1. Select 1 element at at time from left to right
2. Insert the element in its proper position such that all the elements on the left are lesser than this element.
3. Stable Algorithm

11 20 2 9 10 5  
2 11 20 9 10 5  
2 9 11 20 10 5  
2 9 10 11 20 5  
2 5 9 10 11 20  

Time Complexity: O(N^2) worst; O(n) best
Space Complexity: O(1)

````python
static void InsertionSort(int[] arr)
{
    int temp = 0;
    int position = 0;
    for (int i = 1; i < arr.Length; i++)
    {
        temp = arr[i];
        position = i; 
        while (position > 0 && arr[position-1] > temp)
        {
            arr[position] = arr[position-1];
            position--;
        }
        arr[position] = temp;
    }
}

static void Main(string[] args)
{
    int[] arr = { 3, 5, 8, 9, 6, 2 };
    InsertionSort(arr);
    Console.WriteLine("InsertionSort array: ");
    foreach(int i in arr) Console.Write(i + " ");
}
````

#### Shell Sort

1. Select an element and compare it with the element after a gap [ n / 2, n / 3 ]
2. After comparison, do a swap if needed and also compare with left side element after the gap
3. During each pass, we reduce the gap, repeat step 1 and 2.
4. Perform steps 1, 2, 3 till gap becomes 0
<break>

Example:  
Round 1: 3_5_8_9 6 2  
gap (g "_") = 3; 3 < 9 no swap  
Round 2: 3 5_8_9_6; 5 < 6 no swap  
Round 3: 3 5 8_9_6_2; 8 > 2, swap  
Swap: 3 5 2 9 6 8  
cannot check 9 because out of bound, now reduce gape next iteration
<break>

g = 3/2 = 1
Round 4: 3_5 2 9 6 8; 3 < 5 no swap  
Round 5: 3 5_2 9 6 8; 5 > 2 swap  
---------3_2 5 9 6 8; check right side during this swap 3 > 2, swap  
Round 6: 2 3 5_9 6 8; 5 < 9 no swap;  
Round 7: 2 3 5 9_6 8; 9 > 6 swap, 5 < 6 it stays in its swap position  
Round 8: 2 3 5 6 9_8; 9 > 8, swap, check make sure left side less than it and it is  
Round 9: 2 3 5 6 8 9
<break>

gap = gap/2 = 1/2 = 0.5 = 0 => loop complete

````python
static void ShellSort(int[]arr)
{
    int gap = arr.Length/2;
    int i, j;
    while (gap > 0)
    {
        i = gap;
        while (i < arr.Length)
        {
            int temp = arr[i]; // number to be placed in corrent position
            j = i - gap;
            while(j >=0 && arr[j] > temp)
            {
                arr[j+gap] = arr[j];
                j = j-gap;
            }
            arr[j+gap] = temp;
            i++;
        }
        gap = gap/2;
    }
}

static void Main(string[] args)
{
    int[] A = { 3, 5, 8, 9, 6, 2};
    ShellSort(A);
    foreach(var i in A)
    {
        Console.Write(i + " ");
    }
}
````

#### Merge Sort

1. The list is broken into two sub-lists.
2. Keep dividing the sub-lists recursively into two halves until each element is a sub-list.
3. Start merging the lists - starting with combining two sub-lists, which provides a total of four sub-lists. Each element in these sub-lists is now sorted.
4. Keep merging until you get a list with sorted elements.

Advantages
1. Stability : Merge sort is a stable sorting algorithm, which means it maintains the relative order of equal elements in the input array.
2. Guaranteed worst-case performance: Merge sort has a worst-case time complexity of O(N logN) , which means it performs well even on large datasets.
3. Simple to implement: The divide-and-conquer approach is straightforward.
4. Naturally Parallel : We independently merge subarrays that makes it suitable for parallel processing.
<break>

Disadvantages
1. Space complexity: Merge sort requires additional memory to store the merged sub-arrays during the sorting process.
2. Not in-place: Merge sort is not an in-place sorting algorithm, which means it requires additional memory to store the sorted data. This can be a disadvantage in applications where memory usage is a concern.
3. Merge Sort is Slower than QuickSort in general as QuickSort is more cache friendly because it works in-place.
<break>

Time complexity: O(nlogn)  
Space complexity: O(n)  

````python
static void MergeSort(int[] arr, int left, int right)
    {
        if (left < right) //if there are more than one element in the array
        {
            int mid = (left + right) / 2; // find the middle indx
            MergeSort(arr, left, mid); // recursively sort the list in the left half
            MergeSort(arr, mid + 1, right); // recursively sort the list in the right half
            Merge(arr, left, mid, right); // merge the sorted halves
        }
    }

    static void Merge(int[] arr, int left, int right, int mid)
    {
        int[] temp = new int[right - left + 1]; // create a temp array to hold the merged element
        int i = left, j = mid + 1, k = 0; // initialized indices for left and right subarrays and the temp array

        while (i <= mid && j <= right) // while both subarrays have elements
        {
            if (arr[i] <= arr[j]) // compare the elements at the current indices
            {
                temp[k++] = arr[i++]; // add the smaller element to the temporary array and increment the index
            }
            else
            {
                temp[k++] = arr[j++];
            }
        }

        while (i <= mid) // add any remaining elements from the left subarray
        {
            temp[k++] = arr[i++];
        }
        while (j <= right) // add any remaining elements from the right subarray
        {
            temp[k] = arr[j++];
        }
        for (i = left; i <= right, i++) // copy the elements from temp to original array
        {
            arr[i]=temp[i-left];
        }
    }
````
